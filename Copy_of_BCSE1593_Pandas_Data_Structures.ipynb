## 7. Selecting one value from a Series

There are two important ways to select a value.

### `.loc`

`.loc` uses the **label**.

If the index contains:

```text
Aarav, Diya, Kabir, Meera
```

then the label `"Diya"` can be used with `.loc`.

### `.iloc`

`.iloc` uses the **integer position**.

The first position is `0`, the second is `1`, and so on.

Remember:

> `.loc` → label  
> `.iloc` → integer position

### Live coding

Using `student_marks`:

1. Select Diya's marks using `.loc`.
2. Select Diya's marks again using `.iloc`.

````python
print("Diya's marks using label:")

# Write the .loc expression
print(student_marks.loc["Diya"])

print("Diya's marks using position:")

# Write the .iloc expression
print(student_marks.iloc[1])
````

## 8. Selecting multiple values

Both `.loc` and `.iloc` can be used to select more than one value.

With `.loc`, provide a list of labels.

With `.iloc`, provide a list of integer positions.

### Live coding

Using `student_marks`:

- select Aarav and Meera using labels
- select the first and fourth positions

````python
print("Selected using labels:")
# Write the .loc expression
print(student_marks.loc[["Aarav","Meera"]])

print("Selected using positions:")
# Write the .iloc expression
print(student_marks.iloc[[0,3]])
````

## 9. Slicing a Series

Slicing allows us to select a range of values.

For position-based slicing, `.iloc` follows normal Python slicing rules.

For label-based slicing, `.loc` works with labels.

### Live coding

Using `student_marks`:

1. Select the first three students using `.iloc`.
2. Select Aarav through Kabir using `.loc`.

````python
print("First three students:")
# Write the .iloc slicing expression
print(student_marks.iloc[0:3])

print("Aarav through Kabir:")
# Write the .loc slicing expression
print(student_marks.loc["Aarav":"Kabir"])
````

A Series can be used in a condition.

The condition creates a Boolean Series:

Pandas can then use that Boolean result to select the required values.

## 13. Series alignment

One of the useful features of pandas is that labelled objects can be aligned using their index.

Consider two Series:

```text
Quiz 1
Aarav    18
Diya     15
Kabir    20

Quiz 2
Kabir    17
Aarav    19
Diya     16
```

The order is different.

When the Series are added, pandas matches the students by their **labels**, not simply by their positions.

This is called **index alignment**.

### Live coding

Create `quiz1` and `quiz2` with the values above.

Then add them.

Observe the result carefully.

````python
print("Quiz 1:")
# Create quiz1
quiz1=pd.Series([18,15,20],index=["Aarav","Diya","Kabir"])
print(quiz1)

print("Quiz 2:")
# Create quiz2
quiz2=pd.Series([17,19,16],index=["Kabir","Aarav","Diya"])
print(quiz2)

print("Total:")
# Add the two Series
print(quiz1+quiz2)
````

A useful mental model is:

# Part D – Index

## 29. What is an Index?

The index identifies the rows of a pandas object.

If a DataFrame is displayed as:

```text
   Name    Marks
0  Aarav   85
1  Diya    91
2  Kabir   76
3  Meera   88
```

then `0, 1, 2, 3` form the row index.

The index can also contain meaningful labels such as:

```text
E101
E102
E103
```

Indexes are important because pandas uses them for:

- selecting rows
- identifying records
- aligning labelled data
